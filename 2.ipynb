```python
import matplotlib.pyplot as plt

x = ['Mon', 'Tues', 'Wed', 'Thur', 'Fri']
y = [10, 15, 7, 20, 12]

plt.plot(x, y)

plt.title("Bakery Sales This Week!")
plt.xlabel("Day of the Week")
plt.ylabel("Sales Per Day")

plt.grid(True)

# plt.savefig("line_plot.png", dpi=300, bbox_inches="tight")

plt.show()
```
