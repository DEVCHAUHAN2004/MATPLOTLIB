```python
import matplotlib.pyplot as plt 

product = ['A','B','C','D']
sales = [1000,1500,800,1200]

plt.bar(product,sales,color = 'black', label = 'Sales 2025')

plt.savefig("4.png", dpi=300, bbox_inches="tight")

plt.xlabel('Product')
plt.ylabel('Sales')


# plt.xlim(0,1600)


plt.title('Product sales analysis')

plt.legend()

plt.show()



```
